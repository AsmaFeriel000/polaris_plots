```python
from rdkit import Chem
from rdkit.Chem import AllChem
import pandas as pd
```

```python
file = 'sars2_mpro.csv'

#df = pd.read_csv(file, sep=',', skiprows=2, header=None)
df = pd.read_csv(file, sep=',')
df.columns = ["SMILES","Protein Label"]
```

```python
len(df)
```

```python
smi = df["SMILES"][97]

mol = Chem.MolFromSmiles(smi)
molh = Chem.AddHs(mol)


AllChem.EmbedMolecule(molh)
mol = Chem.RemoveHs(molh)
mol
```

```python
sdfFile = open('lig-3D.sdf', 'a')
w = Chem.SDWriter(sdfFile)

for i in range(98):
    smi = df["SMILES"][i]

    mol = Chem.MolFromSmiles(smi)
    molh = Chem.AddHs(mol)
    AllChem.EmbedMolecule(molh)
    mol = Chem.RemoveHs(molh)
    w.write(mol)
```
