# Analysis of Polaris Pose Results

```python
from rdkit import Chem
from rdkit.Chem import AllChem
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

### Compute Competition Metrics and Check We Agree

```python
submitted_poses = Chem.SDMolSupplier('final_submission.sdf')
test_poses = Chem.SDMolSupplier('released_test_poses.sdf')
rmsds = np.array([Chem.rdMolAlign.CalcRMS(test_pose, submitted_pose) for test_pose, submitted_pose in zip(test_poses, submitted_poses)])
```

```python
print(f"Mean RMSD: {rmsds.mean():.2f} Å")
# Get percentage of rmsds below 2.0 Å
success_rate = (rmsds < 2).sum() / len(rmsds) * 100
print(f"Success rate: {success_rate:.2f}%")
```
