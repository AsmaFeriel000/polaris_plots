```python
import polaris as po

competition = po.load_competition("asap-discovery/antiviral-ligand-poses-2025")
```

```python
competition.cache()
```

```python
train, test = competition.get_train_test_split()
```

```python
print(len(train), len(test), len(competition))
```

```python
x, y = train[0]
```

```python
y
```

```python
smiles_list_train = []

for i in range(len(train)):
    x, y = train[i]
    smiles = x["CXSMILES"]
    smiles_list_train.append(smiles)


with open("smiles-train.txt", "w") as f:
    for i in range(len(train)):
        f.write(smiles_list_train[i] + "\n")
```

```python
import fastpdb
import datamol as dm

from tempfile import TemporaryDirectory
from pathlib import Path

for i in range(len(train)):
    datapoint = competition[i]
    
    # The RDKit Mol object supports various IO formats. 
    # Here we use the Datamol library (which wraps RDKit) to effortlessly write to SDF
    filename='ligand-train/mol-{0}.sdf'.format(i)
    dm.to_sdf(datapoint["Ligand Pose"], filename)

    # Similarly, the Biotite AtomArray also support various IO formats
    # Here we use fastpdb, a fast drop-in replacement for Biotite IO.
    out_file = fastpdb.PDBFile()
    out_file.set_structure(datapoint["Complex Structure"])
    filename='complex-train/complex-{0}.pdb'.format(i)
    out_file.write(filename)
```

```python
from rdkit import Chem
from rdkit.Chem import Draw

smiles_list = []
mol_list = []
label_list = []

for i in range(len(test)):
    smiles = test[i]["CXSMILES"]
    label_list.append(test[i]['Protein Label'])
    mol_list.append(Chem.MolFromSmiles(smiles))
    smiles_list.append(smiles)
```

```python
img = Draw.MolsToGridImage(mol_list, molsPerRow=10, returnPNG=False, maxMols=999999, legends=label_list)
img.save('testset.png')
```

```python
with open("smiles-test-SARS.txt", "w") as f:
    for i in range(len(test)):
        if label_list[i] == "SARS-CoV-2 Mpro":
            f.write(smiles_list[i] + "\n")
            #print(i)

with open("smiles-test-MERS.txt", "w") as f:
    for i in range(len(test)):
        if label_list[i] == "MERS-CoV Mpro":
            f.write(smiles_list[i] + "\n")
            #print(i)

```
