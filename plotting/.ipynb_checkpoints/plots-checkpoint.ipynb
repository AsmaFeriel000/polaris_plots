## Figure 2:

```python
# ---------------------------------- sars manual core selection (prospective submission) ----------------
```

```python
# Load molecules
submitted_poses = Chem.SDMolSupplier('../MCS-retrospective/default_fegrow/submitted_default.sdf', removeHs=False)
test_poses = Chem.SDMolSupplier('../released-receptor-structures/released_test_molecules/test_poses.sdf', removeHs=False)

# Load dataset and labels
FIRST_TEST_IDX = 770
ds = DatasetV2.from_json("antiviral-ligand-poses-2025-unblinded/antiviral-ligand-poses-2025-unblinded.json")
ds.load_to_memory()
labels = [ds[i]["Protein Label"] for i in range(FIRST_TEST_IDX, ds.size()[0])]

# Filter MERS molecules
mers_indices = [i for i, label in enumerate(labels) if label == "SARS-CoV-2 Mpro"]

# Create writers
writer_submitted = Chem.SDWriter("submitted_sars.sdf")
writer_test = Chem.SDWriter("test_sars.sdf")

for i in mers_indices:
    submitted_mol = submitted_poses[i]
    test_mol = test_poses[i]
    if submitted_mol is not None:
        writer_submitted.write(submitted_mol)
    if test_mol is not None:
        writer_test.write(test_mol)

# Close writers
writer_submitted.close()
writer_test.close()
```

```python
submitted_poses = Chem.SDMolSupplier('submitted_sars.sdf')
test_poses = Chem.SDMolSupplier('test_sars.sdf')
rmsds = np.array([Chem.rdMolAlign.CalcRMS(test_pose, submitted_pose) for test_pose, submitted_pose in zip(test_poses, submitted_poses)])
```

```python
rmsds
```

```python
len(rmsds)
```

```python
# Zip together the numbers and the rmsds
rmsds_with_numbers = [(i+1, float(rmsd)) for i, rmsd in enumerate(rmsds) if not np.isnan(rmsd)]

# Sort by rmsd
rmsds_with_numbers.sort(key=lambda x: x[1])
```

```python
df = pd.DataFrame(rmsds_with_numbers, columns=["Pymol_Index", "RMSD"])
df.to_csv("sorted_sars_rmsds_submitted_and_released_test.csv", index=False) # then remove the single outlier from the csv
```

```python

# Load the CSV files

df1 = pd.read_csv("sorted_sars_rmsds_submitted_and_released_test.csv")  # remove the outlier
df2 = pd.read_csv("sorted_mcs_sars_rmsds_submitted_and_released_test.csv") 
df3 = pd.read_csv("sorted_MULTITARGET_mcs_sars_rmsds_submitted_and_released_test.csv") # remove the outlier

# Sort RMSD values for CDF calculation

rmsd1 = np.sort(df1["RMSD"].values)
rmsd2 = np.sort(df2["RMSD"].values)
rmsd3 = np.sort(df3["RMSD"].values)

# Compute CDFs
cdf1 = np.arange(1, len(rmsd1) + 1) / len(rmsd1)
cdf2 = np.arange(1, len(rmsd2)+ 1)/ len(rmsd2)
cdf3 = np.arange(1, len(rmsd3)+ 1)/ len(rmsd3)

# Plot CDF vs RMSD
plt.figure(figsize=(10, 6))

plt.plot(rmsd1, cdf1, label="Manual core selection", lw=4)
plt.plot(rmsd2, cdf2, label="MCS-based", lw=4)
plt.plot(rmsd3, cdf3, label="MCS-based (Retrospective)", lw=4)

# Bigger labels
plt.xlabel(r"RMSD ($\AA$)", fontsize=24)
plt.ylabel("Cumulative Distribution Function (CDF)", fontsize=18)

# Bigger tick labels
plt.tick_params(axis='both', which='major', labelsize=26)

# Set x-axis ticks to 2, 4, 6, ...
max_rmsd = max(rmsd1.max(), rmsd2.max())
xticks = np.arange(0, int(np.ceil(max_rmsd)) + 1, 2)
plt.xticks(xticks)



plt.xlim(left=0)

plt.xlim(0, np.ceil(max_rmsd))
plt.ylim(0, 1)

# Remove title (just delete or comment it out)
# plt.title("CDF vs RMSD")

#plt.legend(fontsize=13)
#plt.legend(fontsize=15
#           , frameon=True, loc='lower right')

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# ==========================================================
# OVERLAY PLOT: FEgrow-docked into ApoDock receptors vs prospectively submitted 
# ==========================================================

# Load both datasets
df_fegrow = pd.read_csv("../SARS-CoV-2-Mpro/04-Figure_5/rmsd_of_ligand_vs_pocket_docked_into_apodock_receptors/output/complex_rmsd_results-3-residues.csv")
df_sub    = pd.read_csv("../SARS-CoV-2-Mpro/04-Figure_5/rmsd_of_ligand_vs_pocket_submitted_vs_reference/output/complex_rmsd_results-submitted-vs-ref-3-residues.csv")
                        

# --- Extract mol_id for FEgrow ---
def extract_molid_fegrow(filename):
    m = re.search(r"_mol(\d+)\.pdb$", filename)
    return int(m.group(1)) if m else None

df_fegrow["mol_id"] = df_fegrow["docked_file"].apply(extract_molid_fegrow)

# --- Extract mol_id for Submitted ---
def extract_molid_sub(filename):
    m = re.search(r"complex-MERS-mol(\d+)\.pdb$", filename)
    return int(m.group(1)) if m else None

df_sub["mol_id"] = df_sub["docked_file"].apply(extract_molid_sub)

# --- Lowest ligand RMSD per mol group ---
lowest_idx_fegrow = df_fegrow.groupby("mol_id")["ligand_RMSD_A"].idxmin()
lowest_idx_sub    = df_sub.groupby("mol_id")["ligand_RMSD_A"].idxmin()

df_fegrow["is_lowest"] = False
df_sub["is_lowest"]    = False

df_fegrow.loc[lowest_idx_fegrow, "is_lowest"] = True
df_sub.loc[lowest_idx_sub, "is_lowest"]       = True

# ----------------------------------------------------------
# Percentage of BLUE points (FEgrow lowest-per-mol) with POCKET RMSD < 2 Å
# ----------------------------------------------------------
blue_df = df_fegrow[df_fegrow["is_lowest"]]

total_blue = len(blue_df)
blue_below_2 = (blue_df["pocket_RMSD_A"] < 2).sum()

blue_percentage = (blue_below_2 / total_blue) * 100 if total_blue > 0 else 0

print(f"\nBLUE points (FEgrow lowest-per-mol): {total_blue}")
print(f"BLUE points with POCKET RMSD < 2 Å: {blue_below_2}")
print(f"BLUE percentage: {blue_percentage:.2f}%")

# ----------------------------------------------------------
# Plot
# ----------------------------------------------------------
plt.figure(figsize=(7,7))

# FEgrow (gray + blue highlight)
plt.scatter(
    df_fegrow["pocket_RMSD_A"],
    df_fegrow["ligand_RMSD_A"],
    color="lightgray",
    alpha=0.5,
    label="ApoDock receptors"
)

plt.scatter(
    df_fegrow.loc[df_fegrow["is_lowest"], "pocket_RMSD_A"],
    df_fegrow.loc[df_fegrow["is_lowest"], "ligand_RMSD_A"],
    color="blue",
    s=120,
    edgecolor="black",
    label="ApoDock receptors (lowest per mol)"
)

plt.scatter(
    df_sub["pocket_RMSD_A"],
    df_sub["ligand_RMSD_A"],
    color="red",
    s=120,
    edgecolor="black",
    label="Original Submission"
)

plt.xlabel("Pocket RMSD (Å)")
plt.ylabel("Ligand RMSD (Å)")
#plt.title("Ligand vs Pocket RMSD for SARS-CoV-2 Mpro")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
# ------------------------------------ sars figure 6 ------------------
```

```python
csv_file = "../SARS-CoV-2-Mpro/05-Figure_6/output/residue_rmsd.csv"

output_dir = "output"
os.makedirs(output_dir, exist_ok=True)

# ----------------------------
# HISTOGRAM SETTINGS
# ----------------------------

XMIN = 0.0
XMAX = 3.0
BIN_WIDTH = 0.1
BINS = np.arange(XMIN, XMAX + BIN_WIDTH, BIN_WIDTH)

# ----------------------------
# LOAD DATA
# ----------------------------

df = pd.read_csv(csv_file)

residues = sorted(df["residue"].unique())

# ----------------------------
# PLOT
# ----------------------------

fig, axes = plt.subplots(
    1,
    len(residues),
    figsize=(4 * len(residues), 4),
    sharex=True,
    sharey=True,
)

if len(residues) == 1:
    axes = [axes]

for i, res in enumerate(residues):

    ax = axes[i]

    subset = df[df["residue"] == res]

    apo_vals = subset[
        subset["dataset"] == "ApoDock"
    ]["rmsd"].values

    xtal_vals = subset[
        subset["dataset"] == "Experimental"
    ]["rmsd"].values

    label = subset["label"].iloc[0]

    print(f"\n{label}")
    print(f"ApoDock n = {len(apo_vals)}")
    print(f"Experimental n = {len(xtal_vals)}")

    if len(apo_vals) > 1 and len(xtal_vals) > 1:

        stat, p = ks_2samp(apo_vals, xtal_vals)

        print(f"KS statistic = {stat:.3f}")
        print(f"KS p-value = {p:.3e}")

    sns.histplot(
        xtal_vals,
        bins=BINS,
        stat="density",
        alpha=0.3,
        label="Experimental",
        ax=ax,
    )

    sns.histplot(
        apo_vals,
        bins=BINS,
        stat="density",
        alpha=0.3,
        label="ApoDock",
        ax=ax,
    )

    ax.set_xlim(XMIN, XMAX)
    ax.set_title(label)
    ax.set_xlabel("RMSD (Å)")

    if i == 0:
        ax.set_ylabel("Density")

    if i == len(residues) - 1:
        ax.legend()

plt.suptitle(
    "SARS-CoV-2 Mpro Residue-wise Pocket RMSD Distributions",
    y=1.05,
)

plt.tight_layout()

plt.show()
```

```python


"""
Read bootstrap summary CSV files and produce the combined figure.
"""


def load_summary(csv_file):

    N = []
    mean = []
    low = []
    high = []

    with open(csv_file, newline="") as f:

        reader = csv.DictReader(f)

        for row in reader:
            N.append(int(row["N"]))
            mean.append(float(row["mean"]))
            low.append(float(row["lower_95CI"]))
            high.append(float(row["upper_95CI"]))

    return (
        np.array(N),
        np.array(mean),
        np.array(low),
        np.array(high),
    )


def plot_bootstrap_results(
    aposcore_csv="../SARS-CoV-2-Mpro/07-Figure_9/03_bootstrap/output/aposcore_summary.csv",
    gnina_csv="../SARS-CoV-2-Mpro/07-Figure_9/03_bootstrap/output/gnina_summary.csv",
    random_csv="../SARS-CoV-2-Mpro/07-Figure_9/03_bootstrap/output/random_summary.csv",
    output_plot="combined_bootstrap_plot.png",
):

    N, apos_mean, apos_low, apos_high = load_summary(aposcore_csv)
    _, gnina_mean, gnina_low, gnina_high = load_summary(gnina_csv)
    _, random_mean, random_low, random_high = load_summary(random_csv)

    plt.figure(figsize=(7, 5))

    plt.errorbar(
        N,
        apos_mean,
        yerr=[apos_mean - apos_low, apos_high - apos_mean],
        marker="o",
        capsize=4,
        label="Aposcore",
    )

    plt.errorbar(
        N,
        gnina_mean,
        yerr=[gnina_mean - gnina_low, gnina_high - gnina_mean],
        marker="o",
        capsize=4,
        label="GNINA",
    )

    plt.errorbar(
        N,
        random_mean,
        yerr=[random_mean - random_low, random_high - random_mean],
        marker="o",
        capsize=4,
        label="Random",
    )

    plt.xlabel("Top N")
    plt.ylabel("% RMSD < 2 Å")
    plt.title("Top-N Pose Accuracy for MERS-CoV Mpro")

    plt.legend()
    plt.grid(True)
    plt.xticks(np.arange(1, len(N) + 1, 2))

    plt.tight_layout()

    plt.show()




if __name__ == "__main__":
    plot_bootstrap_results()
```
